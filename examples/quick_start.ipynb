# Quick Start

Minimal workflow to run MitoTNT on 4D mitochondrial movies.

```python
import mitotnt
```


## 1. Data Preparation

**Example folder structure:**

- frame_000/frame_000.tif
- frame_001/frame_001.tif
- ...



```python
# Run MitoGraph segmentation (make sure MitoGraph is in your PATH)
mito_segmenter = mitotnt.MitoSegmenter('D:/GitHub/MitoTNT/test_data/mitograph', xy_pxl_size=0.145, z_pxl_size=0.145)
mito_segmenter.run_mitograph_cpu(max_workers=10)

# Inspect results
mito_segmenter.visualize_segmented_data()
```

## 2. Network Tracking

```python
# Extract graphs
skeletonized_mito = mitotnt.SkeletonizedMito(mito_segmenter)
skeletonized_mito.extract_graphs()
```

```python
# Run tracking
network_tracker = mitotnt.NetworkTracker(skeletonized_mito, frame_interval=3.3)
tracked_mito = network_tracker.run()
```

## 3. Post-tracking analyses

```python
# Detect fusion/fission
tracked_mito.detect_fusion_fission()

# Compute motility
tracked_mito.compute_mitochondria_diffusivity()
```

## 4. Visualization

```python
# Initiate visualization
visualization = mitotnt.Visualizer(tracked_mito)

# Generate skeleton + arrows
visualization.generate_chimerax_skeleton()
visualization.generate_tracking_arrows()

# Create ChimeraX session
visualization.visualize_tracking()
```
